```python
column=["crim","zn","tax","black"]
for col in column:
    minimum=min(df[col])
    maximum=max(df[col])
    df[col]=(df[col]-minimum)/maximum-minimum
```

```python
fig,ax=plt.subplots(ncols=7,nrows=2,figsize=(20,10))
index=0
ax=ax.flatten()
for col,val in df.items():
    sns.distplot(val,ax=ax[index])
    index+=1
plt.tight_layout(pad=0.5,w_pad=0.7,h_pad=0.5)
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
scaled_cols=ss.fit_transform(df[column])
scaled_cols=pd.DataFrame(scaled_cols,columns=column)
scaled_cols.head()
```

```python
for col in column:
    df[col]=scaled_cols[col]
    
```

```python
x=df.drop(columns=['medv','rad'])
y=df["medv"]
from sklearn.model_selection import train_test_split,cross_val_score
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,shuffle=True)
```

```python
from sklearn.metrics import mean_squared_error
def model_train(model):
    model.fit(x_train,y_train)
    y_predict=model.predict(x_test)
    cv_score = cross_val_score(model, x, y, scoring='neg_mean_squared_error', cv=5)
    cv_score = np.abs(np.mean(cv_score))
    print("Model Report")
    print("MSE:",mean_squared_error(y_test, y_predict))
    print('CV Score:', cv_score)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model_train(model)
coef = pd.Series(model.coef_, x.columns).sort_values()
coef.plot(kind='bar', title='Model Coefficients')
```

```python
from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor()
model_train(model)
coef = pd.Series(model.feature_importances_, x.columns).sort_values(ascending=False)
coef.plot(kind='bar', title='Feature Importance')
```

```python
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor()
model_train(model)
coef = pd.Series(model.feature_importances_, x.columns).sort_values(ascending=False)
coef.plot(kind='bar', title='Feature Importance')
```

```python
from sklearn.ensemble import ExtraTreesRegressor
model = ExtraTreesRegressor()
model_train(model)
coef = pd.Series(model.feature_importances_, x.columns).sort_values(ascending=False)
coef.plot(kind='bar', title='Feature Importance')
```
